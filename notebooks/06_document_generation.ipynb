## Drafting Professional Legal Documents using Groq

### Documents we will generate:
1. Rental Agreement
2. Non Disclosure Agreement (NDA)
3. Employment Contract
4. Service Agreement
5. Partnership Deed

### How it works:
User provides → Party names + Terms + Document type
Groq generates → Complete professional legal document
System saves  → Document to outputs/ folder

```python
# Cell 2
import os
import time
from groq import Groq
from dotenv import load_dotenv

# Load environment
load_dotenv('../.env', override=True)
GROQ_API_KEY = os.getenv('GROQ_API_KEY')

# Connect Groq
groq_client = Groq(api_key=GROQ_API_KEY)

print(f"✅ Groq connected!")
print(f"🔑 Key: {GROQ_API_KEY[:8]}...{GROQ_API_KEY[-4:]}")
print()
print("📝 Document Generation System Ready!")
```

## Document Generation Engine
Core function that generates any legal document
Takes document type, parties and terms as input
Returns complete professional legal document

```python
# Cell 4
# Core document generation engine
def generate_document(doc_type, party1, party2, terms, 
                      extra_details=""):
    print(f"📝 Generating: {doc_type}")
    print(f"   Party 1 : {party1}")
    print(f"   Party 2 : {party2}")
    print(f"   Terms   : {terms[:60]}...")
    print()
    
    prompt = f"""You are an expert Indian legal document drafter
with 20 years of experience.

Draft a complete professional {doc_type} under Indian law.

DETAILS:
Party 1        : {party1}
Party 2        : {party2}
Key Terms      : {terms}
Extra Details  : {extra_details}

DOCUMENT STRUCTURE:
1. Document Title
2. Date and Place
3. Parties Section
   - Full names and addresses
   - Roles (landlord/tenant, employer/employee etc)
4. Recitals/Background (WHEREAS clauses)
5. Terms and Conditions
   - At least 8-10 detailed clauses
   - Payment terms
   - Duration and renewal
   - Rights and obligations of each party
6. Termination Clause
   - Notice period
   - Grounds for termination
7. Dispute Resolution
   - Arbitration clause
   - Jurisdiction (mention specific Indian city)
8. Governing Law
   - Mention applicable Indian Acts
9. Force Majeure Clause
10. Signature Section
    - Both parties signatures
    - Witness signatures
    - Date and place

Make it legally sound, professional and 
complete under Indian law.
Use proper legal language."""

    response = groq_client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system",
             "content": "You are an expert Indian legal "
                        "document drafter. Always create "
                        "complete, professional documents."},
            {"role": "user",
             "content": prompt}
        ]
    )
    
    document = response.choices[0].message.content
    
    # Save document
    filename = f"../outputs/{doc_type.replace(' ', '_')}.txt"
    with open(filename, 'w', encoding='utf-8') as f:
        f.write(f"DOCUMENT TYPE: {doc_type}\n")
        f.write(f"PARTY 1: {party1}\n")
        f.write(f"PARTY 2: {party2}\n")
        f.write(f"TERMS: {terms}\n")
        f.write("=" * 60 + "\n\n")
        f.write(document)
    
    print(f"✅ Document generated successfully!")
    print(f"📁 Saved to: {filename}")
    print(f"📊 Length: {len(document)} characters")
    print()
    
    return document, filename

print("✅ Document generation engine ready!")
```

## Challenges & Solutions in 06_document_generation.ipynb

---

### Challenge 1 — Document Quality with Gemini
**Problem:**
- Gemini quota exhausted during document generation
- Documents generated were incomplete
- Missing important legal clauses

**Solution:**
- Switched completely to Groq LLaMA 3.3 70B
- Groq generated complete 5000+ character documents
- All legal clauses included automatically

**Lesson Learned:**
- Groq is better for long document generation
- LLaMA 3.3 70B understands Indian law very well
- Always specify document structure in prompt

---

### Challenge 2 — Missing Legal Clauses
**Problem:**
- Simple prompts generated basic documents
- Missing force majeure, arbitration, jurisdiction
- Not professional enough for real use

**Solution:**
- Added detailed document structure in prompt
- Specified exactly 10 sections to include
- Added Indian Acts references in extra_details
- Result: 5000+ character professional documents

**Lesson Learned:**
- Prompt engineering is critical for quality
- More specific prompt = better document
- Always mention governing Indian Acts in prompt

---

### Challenge 3 — File Naming Issues
**Problem:**
- Document type with spaces caused file naming issues
- "Non Disclosure Agreement" → filename had spaces
- Some operating systems don't handle spaces well

**Solution:**
- Used .replace(' ', '_') for all filenames
- "Non Disclosure Agreement" → "Non_Disclosure_Agreement.txt"
- Clean filenames work on all operating systems

**Lesson Learned:**
- Always sanitize filenames before saving
- Replace spaces with underscores
- Avoid special characters in filenames

---

### Challenge 4 — Document Preview Truncated
**Problem:**
- Full documents were 5000+ characters
- VS Code cell output truncated long text
- Could not see complete document in notebook

**Solution:**
- Used [:1000] to show only first 1000 chars
- Full document saved to outputs/ folder
- Open .txt file directly to see complete document

**Lesson Learned:**
- Never display full large text in notebook cells
- Always save to file and show preview only
- Use VS Code to open generated .txt files
