# 01 — Data Exploration
## Understanding our 30,444 Indian Legal Sections
Before building any AI, we must understand our data first.

```python
import pandas as pd

# Load the Excel file
df = pd.read_excel('../data/Scrapmetadata_forvectordatabase.xlsx')

print("✅ File loaded successfully!")
print(f"📊 Total rows: {len(df)}")
print(f"📋 Total columns: {len(df.columns)}")
```

## Step 1 — See All Column Names
What information do we have for each law section?

```python
# See all column names
print("📋 All columns in our dataset:")
print()
for i, col in enumerate(df.columns):
    print(f"  Column {i+1}: {col}")
```

## Step 2 — Preview the Data
See first 5 rows of our dataset

```python
# See first 5 rows
df.head()
```

## Step 3 — Count Unique Acts
How many different Indian Laws are in our dataset?

```python
# Count unique Acts
print(f"📚 Total unique Acts in dataset: {df['Act Title'].nunique()}")
print()
print("📋 Top 10 Acts with most sections:")
print()
print(df['Act Title'].value_counts().head(10))
```

## Step 4 — Check Missing Values
Find which columns have empty/missing data

```python
# Check missing values in each column
print("🔍 Missing values in each column:")
print()
missing = df.isnull().sum()
for col, count in missing.items():
    status = "✅ No missing" if count == 0 else f"⚠️ {count} missing"
    print(f"  {col}: {status}")
```

## Step 5 — Check Content Length
How long are our law sections? Short or long?

```python
# Check length of legal content
df['content_length'] = df['Content'].str.len()

print("📏 Content Length Statistics:")
print()
print(f"  Shortest section : {int(df['content_length'].min())} characters")
print(f"  Longest section  : {int(df['content_length'].max())} characters")
print(f"  Average length   : {int(df['content_length'].mean())} characters")
print()
print("📊 Length distribution:")
print(f"  Under 500 chars  : {len(df[df['content_length'] < 500])} sections")
print(f"  500 - 1000 chars : {len(df[(df['content_length'] >= 500) & (df['content_length'] < 1000)])} sections")
print(f"  Above 1000 chars : {len(df[df['content_length'] > 1000])} sections")
```

## Step 6 — Read One Full Legal Section
See exactly what one law section looks like

```python
# Read one full legal section
print("📖 Sample Legal Section:")
print("=" * 60)
sample = df[df['Content'].notna()].iloc[0]
print(f"Act Title       : {sample['Act Title']}")
print(f"Act ID          : {sample['Act ID']}")
print(f"Section ID      : {sample['Section ID']}")
print(f"Section Heading : {sample['Section Heading']}")
print(f"Chapter Name    : {sample['Chapter Name']}")
print(f"Enactment Date  : {sample['Enactment Date']}")
print()
print("📝 Full Content:")
print("-" * 60)
print(sample['Content'])
```
